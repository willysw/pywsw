```python
from pywsw.analysis.foundations import Overturning
from pywsw.core.utility import parse_as_ndarray
import numpy as np
```

```python
base_args = {
    'toe_length':               (8/12, 6.0),
    'heel_length':              (8/12, 1.0),
    'height':                   (1.0, 2.0),
    'width':                    1.0,
    'bearing_length':           8/12,
    'toe_super_unit_load':      150.0,
    'heel_super_unit_load':     9*150.0,
    'unit_weight':              150.0,
    'superimposed_load':        150*(8/12)*10,
}
```

```python
rws = [
    Overturning.optimize(overturning_moment=mi, min_safety_factor=sfi, allow_bearing_pressure=qai, **base_args)
    for mi, sfi, qai in
    ((5670.0, 1.5, 1500.0), (6150, 1.1, 2000.0))
]

for rwi in rws:
    print(f"{rwi.toe_length=}, {rwi.heel_length=}, {rwi.safety_factor=}, {rwi.max_bearing_pressure=}")
```
